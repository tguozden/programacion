```python
# Reglas de visibilidad de variables

# 1:  scope actual
# 2:  scope superior
# 3:  módulos utilizados
# 4:  módulos por defecto
```

```python
def imprime_valor():
    print('x vale',x)
try:
    imprime_valor()
except Exception as e:
    print(f"error: {e}")
```

```python
# aquí x no está en el scope actual
# se busca en el scope superior
x = 1.23
imprime_valor()
```

```python
# aquí var está en el scope actual
# hay dos variables que se llaman var
var = 2
def imprime_var():
    var = 1
    print('la función imprime', var)

imprime_var()
print('afuera queda', var)
```

```python
# los scopes dentro de funciones esconden o "encierran" a las variables
# (closures)
def hola():
    z = 3
    print('hola', z)
hola()

try:
    print('afuera', z)
except Exception as e:
    print(f"Hubo un error: {e}")
```

```python
# if/for/while/with no definen nuevos scopes:

for i in range(3):
    pass
print(i)
```

```python
# en cambio las comprensiones sí!!
# por ejemplo las listas por comprensión
del j
print([j for j in range(3)])
# for j in range(3):
#     print(j)
try:
    print('el valor de j quedó', j)
except Exception as e:
    print(f"Hubo un error: {e}")
# las variables definidas dentro
# del sintaxis por comprensión
# no persisten
```

```python
# variables globales
del bananas
```

```python
bananas = 'no tienen carozo'
```

```python
def funcion1():
    bananas = 'tienen filamento' # esta asignación no afecta a la variable global

def funcion2():
    global bananas
    bananas = 'tienen filamento groso' # esta asignación no afecta a la variable global
```

```python
funcion1()
print(bananas)
```

```python
funcion2()
print(bananas)
```

```python
contador = 0  # Variable global

def incrementar():
    global contador  # Referencia a la variable global
    contador += 1

incrementar()
print(contador)
```

```python
# ver que acá contador no se modifica
contador = 0

def intentar_incrementar():
    contador = 1  # variable local: no modifica del ámbito global

intentar_incrementar()
print(contador)

```

```python
# a continuación esto falla:

contador = 0  # Variable de la función exterior

def funcion():
    nonlocal contador # contador debería estar en un scope superior pero no el ámbito global
    contador += 1
    return contador
```

```python
contador = 'este nada que ver'
def funcion_exterior():
    contador = 0  # variable en scope superior o exterior
   
    def funcion_interior():
        nonlocal contador  # Referencia al scope inferior o interior
        contador += 1
        return contador
    
    return funcion_interior

mi_funcion = funcion_exterior()
print(mi_funcion())
```

# Algunos consejos
- Da prioridad a las funciones pequeñas y específicas.
    - Define las variables en el ámbito más restringido posible.
- Pasar datos entrantes y salientes a través de parámetros y valores de retorno. Minimizar el estado a nivel global.
- Utiliza los scopes con atención: toma nota de qué nombres externos capturan la función anidada. Usar nonlocal cuando realmente se necesite volver a vincular una variable.
- Eligir nombres descriptivos, que no den lugar a conflictos. El guión bajo inicial (por ejemplo, _cache) indica uso interno.
    - por ej. variables enteras suelen comenzar con i, j, k, l, n, m.
    - Pascal Casing
