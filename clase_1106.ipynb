```python
# qué pasa si mandamos la variable de una función como argumento?
# pasando funciones
def hola_mundo():
    return 'hola mundo'

def toma_funcion(func):
    cadena = func()
    print('toma funcion me dio ',cadena)

toma_funcion(hola_mundo)
    
```

```python
# se puede usar para aplicar funciones en una lista
def aplica_funcion(func, lista):
    list_x_comp = [func(elemento) for elemento in lista]
    return list_x_comp

def por2(x):
    return x * 2

def cuadrado(x):
    return x ** 2

numeros = [1, 2, 3, 4, 5]
aplica_funcion(por2, numeros)
```

```python
# recordatorio!! aplica funcion es parecido a map como habíamos visto antes
# solo que map crea un iterador: espera para ser ejecutado (como range)
map(por2, numeros)
```

```python
list(map(por2, numeros))
```

```python
# otro recordatorio: las funciones sencillas como por2() podían implementarse con lambda:
otra_por2 = lambda x: 2*x
otra_por2(4)
```

```python
# y si las funciones devuelven funciones
def multi_por_factor(factor):
    #
    #
    
    def multi(x):
        return x * factor
    return multi

# creando funciones distintas
multi_por_factor(2)
```

```python
multi_por_factor(2)(7)
```

```python
duplicar = multi_por_factor(2)
triplicar = multi_por_factor(3)

print(duplicar(5))
print(triplicar(5))
```

```python
# algo que pasó y que no nos dimos cuenta

def exterior(x):
    def interior(y):
    # el valor de x queda en la función interior
        return x + y 
    return interior


closure = exterior(10) 
# aquí el valor de x queda definido cuando creamos la instancia de la función


print(closure(5)) 
print(closure(20))
```

```python
def make_counter():
    count = 0  # esta es la variable que qeuda "encerrada"
    def counter():
        nonlocal count  # toma y modifica la variable del scope exterior
        count += 1
        return count
    return counter

counter1 = make_counter()
```

```python
counter1()
```

```python
print(counter1())  # 1
print(counter1())
```

```python
def func_compuesta(func):
    def disfraz():
        nuevomensaje = func().upper()
        return nuevomensaje
    return disfraz
def holamundo():
    return 'hola, mundo!'        
funcion_decorada = func_compuesta(holamundo)

funcion_decorada()
```

```python
@func_compuesta
def holamundo():
    return 'hoy es martes'
    
holamundo()
```

```python
def cachear(func):
    cache = {}
    
    def wrapper(*args):
        if args in cache:
            print("Resultado desde cache")
            return cache[args]
        resultado = func(*args)
        cache[args] = resultado
        print("Resultado calculado")
        return resultado
    return wrapper

@cachear
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
print(factorial(5))  # Esta llamada usa el cache
```
